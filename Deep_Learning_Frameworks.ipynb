##Q2. How can you define a simple function in TensorFlow 2.0 to perform addition?

````python
@tf.function
def addition(a, b):
    return a + b

a = tf.constant(10)
b = tf.constant(20)

result = addition(a, b)

print("Addition Result:", result)
print("Addition Result:", result.numpy())
````

Explanation:
- `tf.constant()` creates TensorFlow tensors.
- `@tf.function` converts the Python function into a graph-compatible TensorFlow function.
- `a + b` performs tensor addition.
- `.numpy()` converts the result into a NumPy value.

##Q3. How can you create a simple neural network in TensorFlow 2.0 with one hidden layer?

````python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

# Generate sample data
X = np.random.rand(500, 2)
y = (X[:, 0] + X[:, 1] > 1).astype(int)

# Create neural network
model = Sequential([
    Dense(8, activation="relu", input_shape=(2,)),
    Dense(1, activation="sigmoid")
])

# Compile model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Train model
model.fit(X, y, epochs=20, batch_size=32)

# Evaluate model
loss, accuracy = model.evaluate(X, y)

print("Loss:", loss)
print("Accuracy:", accuracy)
````

Explanation:

The neural network contains:
1. Input layer: Accepts two features.
2. Hidden layer: Contains eight neurons with ReLU activation.
3. Output layer: Contains one neuron with sigmoid activation.
The model learns to classify observations into two classes.

##Q4. How can you visualize the training progress using TensorFlow and Matplotlib?

````python
import matplotlib.pyplot as plt

# Generate sample data
X = np.random.rand(500, 2)
y = (X[:, 0] + X[:, 1] > 1).astype(int)

# Create model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(
        8, activation="relu", input_shape=(2,)
    ),
    tf.keras.layers.Dense(1, activation="sigmoid")
])

# Compile model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Train model
history = model.fit(
    X, y,
    epochs=20,
    validation_split=0.2,
    verbose=1
)

# Plot training and validation loss
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("Model Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()

# Plot training and validation accuracy
plt.subplot(1, 2, 2)
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("Model Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()

plt.tight_layout()
plt.show()
````

Explanation:
- `history` stores the training results.
- `loss` represents training loss.
- `val_loss` represents validation loss.
- `accuracy` represents training accuracy.
- `val_accuracy` represents validation accuracy.

Matplotlib displays the changes in loss and accuracy over the training epochs.

##Q6. How do you create a simple neural network in PyTorch?

````python
import torch.nn as nn

# Create neural network
class NeuralNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(10, 64)
        self.relu = nn.ReLU()
        self.layer2 = nn.Linear(64, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.layer1(x)
        x = self.relu(x)
        x = self.layer2(x)
        x = self.sigmoid(x)

        return x


# Create model
model = NeuralNetwork()

# Generate sample input
X = torch.randn(5, 10)

# Forward propagation
output = model(X)

print("Model Output:")
print(output)

print("Output Shape:", output.shape)
````

Explanation:
- `nn.Module` is the base class for neural networks.
- `nn.Linear()` creates fully connected layers.
- `nn.ReLU()` introduces non-linearity.
- `nn.Sigmoid()` converts outputs into values between 0 and 1.
- `forward()` defines the flow of data through the network.

The network accepts 10 input features and produces one output per sample.

##Q7. How do you define a loss function and optimizer in PyTorch?

####A loss function measures the difference between predicted values and actual values.

An optimizer updates the network parameters to minimize the loss.

**Example:**

````python
# Create model
model = nn.Linear(10, 1)

# Define loss function
criterion = nn.MSELoss()

# Define optimizer
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

# Generate sample data
X = torch.randn(100, 10)
y = torch.randn(100, 1)

# Forward propagation
predictions = model(X)

# Calculate loss
loss = criterion(predictions, y)

# Backpropagation
optimizer.zero_grad()
loss.backward()

# Update model parameters
optimizer.step()

print("Training Loss:", loss.item())
````

Explanation:
1. `nn.MSELoss()` calculates Mean Squared Error.
2. `torch.optim.Adam()` creates the optimizer.
3. `optimizer.zero_grad()` clears old gradients.
4. `loss.backward()` calculates gradients.
5. `optimizer.step()` updates the model parameters.

```
Common loss functions:

Loss Function       Purpose
MSELoss             Regression
L1Loss              Regression
BCELoss             Binary classification with probabilities
CrossEntropyLoss    Multiclass classification
```



##Q8. How do you implement a custom loss function in PyTorch?

####A custom loss function is created when the standard loss functions do not meet the requirements of a particular problem.

**Example: Custom Mean Absolute Error (MAE)**

````python
def custom_mae(y_true, y_pred):
    loss = torch.mean(torch.abs(y_true - y_pred))
    return loss

# Actual values
y_true = torch.tensor([10.0, 20.0, 30.0])

# Predicted values
y_pred = torch.tensor([12.0, 18.0, 28.0])

# Calculate custom loss
loss = custom_mae(y_true, y_pred)

print("Custom MAE Loss:", loss.item())
````

Explanation:
- `torch.abs()` calculates absolute differences.
- `torch.mean()` calculates their average.
- The function returns the Mean Absolute Error.

Custom loss functions can also be used directly during neural network training, provided their operations support gradient calculation.

##Q9. How do you save and load a TensorFlow model?

####TensorFlow provides built-in methods to save and load trained models.

**Step 1: Create and train a model**

````python
# Generate sample data
X = np.random.rand(500, 2)
y = (X[:, 0] + X[:, 1] > 1).astype(int)

# Create model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(
        8, activation="relu", input_shape=(2,)
    ),
    tf.keras.layers.Dense(1, activation="sigmoid")
])

# Compile model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Train model
model.fit(X, y, epochs=10)
````

**Step 2: Save the model**

````python
model.save("my_model.keras")

print("Model saved successfully!")
````

**Step 3: Load the model**

````python
loaded_model = tf.keras.models.load_model("my_model.keras")

print("Model loaded successfully!")
````

**Step 4: Make predictions**

````python
sample = np.array([[0.8, 0.7]])

prediction = loaded_model.predict(sample)

print("Prediction:", prediction)
````

Explanation:
- `model.save()` saves the complete model, including its architecture, weights, and training configuration.
- `load_model()`restores the saved model.
- `predict()` generates predictions using the loaded model.
Main advantage: Saving and loading allows us to reuse trained models without training them again.
