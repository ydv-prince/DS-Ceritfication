## Q3. Write a Python function to calculate the probability distribution function (PDF) of a Bernoulli distribution.
*Note: A Bernoulli distribution has two possible outcomes: 0 and 1.*

```python
def bernoulli_pmf(x, p):
    if x == 1:
        return p
    elif x == 0:
        return 1 - p
    else:
        return 0

p = 0.6
print("P(X=0):", bernoulli_pmf(0, p))
print("P(X=1):", bernoulli_pmf(1, p))
```

**Explanation:**
- `p` represents the probability of success.
- `1-p` represents the probability of failure.
- The function returns the probability for the specified outcome.

*Note: Technically, this is a PMF, not a continuous PDF, because Bernoulli is a discrete distribution.*

## Q6. Write a Python program to calculate and plot the cumulative distribution function (CDF) of a discrete uniform distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

# Outcomes
x = np.arange(1, 7)
# Probability of each outcome
pmf = np.ones(6) / 6
# Calculate CDF
cdf = np.cumsum(pmf)

# Plot CDF
plt.step(x, cdf, where="post", marker="o")
plt.title("CDF of Discrete Uniform Distribution")
plt.xlabel("Outcomes")
plt.ylabel("Cumulative Probability")
plt.xticks(x)
plt.ylim(0, 1.1)
plt.grid()
plt.show()
print("CDF:", cdf)
```

**Explanation:**
- `pmf` contains equal probabilities.
- `np.cumsum()` calculates cumulative probabilities.
- `plt.step()` displays the cumulative distribution.

## Q9. Write a Python function to calculate Z-scores from a dataset and plot them.

```python
import numpy as np
import matplotlib.pyplot as plt

def calculate_zscore(data):
    mean = np.mean(data)
    std = np.std(data)
    z_scores = (data - mean) / std
    return z_scores

# Dataset
data = np.array([10, 20, 30, 40, 50, 60, 70])
# Calculate Z-scores
z_scores = calculate_zscore(data)
print("Z-Scores:", z_scores)

# Plot
plt.plot(data, z_scores, marker="o")
plt.axhline(0, linestyle="--")
plt.title("Z-Score Visualization")
plt.xlabel("Original Values")
plt.ylabel("Z-Scores")
plt.grid()
plt.show()
```

**Explanation:**
- `np.mean()` calculates the mean.
- `np.std()` calculates the population standard deviation by default.
- The Z-score formula standardizes each observation.
- The graph displays the standardized values.

## Q10. Implement the Central Limit Theorem (CLT) using Python for a non-normal distribution.
*We will use an exponential distribution, which is positively skewed.*

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate non-normal population
population = np.random.exponential(scale=2, size=100000)

# Calculate sample means
sample_means = []
for i in range(1000):
    sample = np.random.choice(population, size=50)
    sample_means.append(np.mean(sample))

# Plot population and sample means
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.hist(population, bins=40, density=True)
plt.title("Exponential Population")

plt.subplot(1, 2, 2)
plt.hist(sample_means, bins=30, density=True)
plt.title("Distribution of Sample Means")
plt.tight_layout()
plt.show()
```

**Explanation:**
- The exponential distribution represents the original non-normal population.
- 1,000 samples are generated.
- Each sample contains 50 observations.
- The mean of every sample is calculated.
- The distribution of sample means becomes approximately normal. This demonstrates the Central Limit Theorem.

## Q15. Implement hypothesis testing using Z-statistics for a sample dataset.
*We will perform a one-sample, two-sided Z-test with a known population standard deviation.*

**Hypotheses:**
- Null hypothesis (H₀): Population mean = 50.
- Alternative hypothesis (H₁): Population mean is not equal to 50.

```python
import numpy as np
from scipy.stats import norm

# Sample data
sample = np.array([52, 54, 49, 51, 53, 55, 50, 52, 54, 51])
# Hypothesized population mean
mu = 50
# Known population standard deviation
sigma = 5

# Sample size and mean
n = len(sample)
sample_mean = np.mean(sample)

# Calculate Z-statistic
z = (sample_mean - mu) / (sigma / np.sqrt(n))

# Calculate two-tailed p-value
p_value = 2 * (1 - norm.cdf(abs(z)))
alpha = 0.05

print("Sample Mean:", sample_mean)
print("Z-Statistic:", z)
print("P-Value:", p_value)

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

**Explanation:**
1. Calculate the sample mean.
2. Calculate the Z-statistic.
3. Calculate the p-value.
4. Compare the p-value with the significance level of 0.05.
5. Make a statistical decision.

*Note: This example assumes the population standard deviation is known. If it is unknown, a one-sample t-test is generally appropriate instead.*

## Q16. Create a confidence interval for a dataset using Python and interpret the result.
*We will calculate a 95% confidence interval using a t-distribution.*

```python
import numpy as np
from scipy.stats import t

# Sample data
data = np.array([45, 50, 52, 48, 55, 53, 49, 51, 54, 47])

# Calculate sample statistics
mean = np.mean(data)
std = np.std(data, ddof=1)
n = len(data)

# Critical t-value
t_value = t.ppf(0.975, df=n - 1)

# Calculate margin of error
margin = t_value * std / np.sqrt(n)

# Confidence interval
lower = mean - margin
upper = mean + margin

print("Sample Mean:", mean)
print("95% Confidence Interval:")
print(lower, "to", upper)
```

**Explanation:**
- `np.mean()` calculates the sample mean.
- `ddof=1` calculates the sample standard deviation.
- `t.ppf()` calculates the critical t-value.
- The confidence interval gives a range of plausible population mean values.

**Interpretation:** We are 95% confident that the interval calculated using this method contains the true population mean, assuming the sampling conditions are satisfied.

## Q22. Calculate the range and interquartile range (IQR) of a dataset generated from a normal distribution.

```python
import numpy as np

# Generate normal distribution data
data = np.random.normal(loc=50, scale=10, size=1000)

# Calculate range
data_range = np.max(data) - np.min(data)

# Calculate quartiles
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

# Calculate IQR
IQR = Q3 - Q1

print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Range:", data_range)
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

**Explanation:**
- Range = Maximum − Minimum.
- `Q1` represents the 25th percentile.
- `Q3` represents the 75th percentile.
- `IQR = Q3 − Q1`.
- The IQR measures the spread of the middle 50% of the data.

## Q23. Implement Z-score normalization on a dataset and visualize its transformation.

```python
import numpy as np
import matplotlib.pyplot as plt

# Original dataset
data = np.array([10, 20, 30, 40, 50, 60, 70])

# Calculate mean and standard deviation
mean = np.mean(data)
std = np.std(data)

# Z-score normalization
normalized_data = (data - mean) / std
print("Original Data:", data)
print("Normalized Data:", normalized_data)

# Visualization
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.hist(data, bins=7, edgecolor="black")
plt.title("Original Data")

plt.subplot(1, 2, 2)
plt.hist(normalized_data, bins=7, edgecolor="black")
plt.title("Z-Score Normalized Data")
plt.tight_layout()
plt.show()
```

**Explanation:**
- Calculate the mean and standard deviation.
- Subtract the mean from each observation.
- Divide each result by the standard deviation.
- The transformed dataset has mean approximately 0 and standard deviation approximately 1.

## Q24. Write a Python function to calculate the skewness and kurtosis of a dataset generated from a normal distribution.

```python
import numpy as np
from scipy.stats import skew, kurtosis

# Generate normal distribution data
data = np.random.normal(loc=50, scale=10, size=10000)

# Calculate skewness
skewness = skew(data)

# Calculate kurtosis
kurt = kurtosis(data, fisher=True)

print("Skewness:", skewness)
print("Kurtosis:", kurt)
```

**Explanation:**
- `skew()` measures the asymmetry of a distribution.
- `kurtosis()` measures tail heaviness relative to a normal distribution.
- `fisher=True` calculates excess kurtosis, for which a normal distribution has a theoretical value of zero.

### Interpretation Guide:

| Measurement | Meaning |
| :--- | :--- |
| **Skewness = 0** | Symmetric distribution |
| **Skewness > 0** | Right-skewed distribution |
| **Skewness < 0** | Left-skewed distribution |
| **Excess kurtosis = 0** | Normal-distribution reference |
| **Excess kurtosis > 0** | Heavier tails than normal |
| **Excess kurtosis < 0** | Lighter tails than normal |
