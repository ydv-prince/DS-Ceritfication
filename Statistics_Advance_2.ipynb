##Question 5: What is the difference between a Z-test and a T-test? Explain when to use each.

````python
import numpy as np
from scipy import stats

# Sample data
data = [12, 14, 15, 16, 18, 20, 22, 24]

# Calculate sample mean
sample_mean = np.mean(data)

# Perform one-sample T-test
t_stat, p_value = stats.ttest_1samp(data, popmean=15)

print("Sample Mean:", sample_mean)
print("T-statistic:", t_stat)
print("P-value:", p_value)
````

##Question 6: Write a Python program to generate a binomial distribution with n=10 and p=0.5, then plot its histogram.

(Include your Python code and output in the code box below.)
Hint: Generate random number using random function.

````python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Define parameters
n = 10
p = 0.5
size = 1000

# Generate binomial distribution
data = np.random.binomial(n, p, size)

# Print generated data
print("First 20 observations:")
print(data[:20])

# Calculate mean and variance
print("Mean:", np.mean(data))
print("Variance:", np.var(data))

# Plot histogram
plt.figure(figsize=(8, 5))

plt.hist(
    data,
    bins=np.arange(-0.5, n + 1.5, 1),
    edgecolor='black',
    color='skyblue'
)

plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")
plt.xticks(range(n + 1))
plt.grid(axis='y', alpha=0.3)

plt.show()
````

##Question 7: Implement hypothesis testing using Z-statistics for a sample dataset in Python. Show the Python code and interpret the results.

```
sample_data = [
49.1, 50.2, 51.0, 48.7, 50.5, 49.8,
50.3, 50.7, 50.2, 49.6, 50.1, 49.9,
50.8, 50.4, 48.9, 50.6, 50.0, 49.7,
50.2, 49.5, 50.1, 50.3, 50.4, 50.5,
50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
50.3, 50.4, 50.0, 49.7, 50.5, 49.9
]
```


(Include your Python code and output in the code box below.)

````python
import numpy as np
from scipy.stats import norm

# Given sample dataset
sample_data = [
    49.1, 50.2, 51.0, 48.7, 50.5, 49.8,
    50.3, 50.7, 50.2, 49.6, 50.1, 49.9,
    50.8, 50.4, 48.9, 50.6, 50.0, 49.7,
    50.2, 49.5, 50.1, 50.3, 50.4, 50.5,
    50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
    50.3, 50.4, 50.0, 49.7, 50.5, 49.9
]

# Define parameters
population_mean = 50
population_std = 1.0  # Assumed known population SD
alpha = 0.05

# Calculate sample statistics
sample_mean = np.mean(sample_data)
n = len(sample_data)

# Calculate Z-statistic
z_stat = (
    (sample_mean - population_mean)
    / (population_std / np.sqrt(n))
)

# Calculate two-tailed p-value
p_value = 2 * (1 - norm.cdf(abs(z_stat)))

# Print results
print("Sample Size:", n)
print("Sample Mean:", round(sample_mean, 4))
print("Population Mean:", population_mean)
print("Z-statistic:", round(z_stat, 4))
print("P-value:", round(p_value, 4))

# Make decision
if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
````

##Question 8: Write a Python script to simulate data from a normal distribution and calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.
(Include your Python code and output in the code box below.)

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Set random seed
np.random.seed(42)

# Generate normal distribution data
data = np.random.normal(
    loc=50,
    scale=10,
    size=100
)

# Calculate sample statistics
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)
n = len(data)

# Calculate standard error
standard_error = sample_std / np.sqrt(n)

# Calculate 95% confidence interval
confidence_level = 0.95

lower, upper = stats.t.interval(
    confidence=confidence_level,
    df=n - 1,
    loc=sample_mean,
    scale=standard_error
)

# Print results
print("Sample Mean:", round(sample_mean, 4))
print("Sample Standard Deviation:", round(sample_std, 4))
print("95% Confidence Interval:")
print("Lower Bound:", round(lower, 4))
print("Upper Bound:", round(upper, 4))

# Plot histogram
plt.figure(figsize=(8, 5))

plt.hist(
    data,
    bins=15,
    density=True,
    edgecolor='black',
    color='skyblue',
    alpha=0.7
)

plt.axvline(
    sample_mean,
    color='red',
    linestyle='--',
    label='Sample Mean'
)

plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Density")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
````

##Question 9: Write a Python function to calculate the Z-scores from a dataset and visualize the standardized data using a histogram. Explain what the Z-scores represent in terms of standard deviations from the mean.
(Include your Python code and output in the code box below.)

````python
import numpy as np
import matplotlib.pyplot as plt

# Define a dataset
data = np.array([10, 20, 30, 40, 50])

# Function to calculate Z-scores
def calculate_z_scores(data):

    # Calculate mean
    mean = np.mean(data)

    # Calculate standard deviation
    std = np.std(data)

    # Calculate Z-scores
    z_scores = (data - mean) / std

    return z_scores

# Calculate Z-scores
z_scores = calculate_z_scores(data)

# Print original data and Z-scores
print("Original Data:", data)
print("Z-Scores:", np.round(z_scores, 4))

# Plot histogram
plt.figure(figsize=(8, 5))

plt.hist(
    z_scores,
    bins=5,
    edgecolor='black',
    color='skyblue'
)

plt.axvline(
    0,
    color='red',
    linestyle='--',
    label='Mean (Z = 0)'
)

plt.title("Histogram of Standardized Data")
plt.xlabel("Z-Scores")
plt.ylabel("Frequency")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
````
