##Q1. How can you open a file for writing in Python and write a string to it?

```python
# Open file in write mode
with open("data.txt", "w") as file:

    # Write string into file
    file.write("Hello, Welcome to Python!")

print("Data written successfully!")
```

Explanation:
- `"w"` opens a file in write mode.
- `write()` writes text into the file.
- If the file does not exist, Python creates it.
- If the file already exists, its previous content is overwritten.
- with automatically closes the file.

##Q2. Write a Python program to read the contents of a file and print each line.

```python
with open("data.txt", "r") as file:

    for line in file:
        print(line, end="")

print("\nFile reading completed!")
```

Explanation:

- The `for` loop reads one line at a time.
- `end=""` prevents extra blank lines caused by existing newline characters.

##Q3. How would you handle a case where the file doesn't exist while trying to open it for reading?

```python
try:
    with open("data.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("Error: File does not exist!")
```

Explanation: If the file is missing, Python executes the except block instead of terminating the program with an unhandled exception.

##Q4. Write a Python script that reads from one file and writes its content to another file.

```python
try:
    # Read source file
    with open("data.txt", "r") as source:
        content = source.read()
    # Write into destination file
    with open("destination.txt", "w") as destination:
        destination.write(content)
    print("File copied successfully!")
except FileNotFoundError:
    print("Source file not found!")
except OSError as e:
    print("File error:", e)
```

Explanation:
1. Open the source file.
2. Read its content.
3. Open the destination file.
4. Write the content into the destination file.
5. Handle file-related errors.

##Q18. How would you implement a basic logging setup that logs to a file with rotation after 1MB?

```python
import logging
from logging.handlers import RotatingFileHandler

# Create rotating file handler
handler = RotatingFileHandler(
    "application.log",
    maxBytes=1024 * 1024,
    backupCount=3
)
# Configure logging
logging.basicConfig(
    level=logging.INFO,
    handlers=[handler],
    format="%(asctime)s - %(levelname)s - %(message)s"
)
# Generate log messages
logging.info("Application started.")
logging.warning("Memory usage is increasing.")
logging.error("An error occurred.")
print("Logging configured successfully!")
```

Explanation:
- `RotatingFileHandler` manages log file rotation.
- `maxBytes=1024*1024` sets the maximum size to 1 MB.
- `backupCount=3` keeps up to three backup files.
- Older log files are rotated when the current file reaches the size limit.

##Q21. Write a Python program that reads a file and prints the number of occurrences of a specific word.

```python
filename = "data.txt"
word = "Python"
try:
    with open(filename, "r") as file:
        content = file.read()
        words = content.split()
        count = sum(
            1 for w in words
            if w.strip(".,!?;:").lower() == word.lower()
        )
        print("Word:", word)
        print("Occurrences:", count)
except FileNotFoundError:
    print("File not found!")
```

Explanation:
- `read()` reads the file content.
- `split()` separates the content into words.
- `strip()` removes common punctuation around words.
- `lower()` makes the comparison case-insensitive.
- `sum()` counts matching words.

##Q22. How can you check if a file is empty before attempting to read its contents?

```python
import os
filename = "data.txt"
try:
    if os.path.getsize(filename) == 0:
        print("File is empty!")
    else:
        with open(filename, "r") as file:
            print(file.read())
except FileNotFoundError:
    print("File does not exist!")
```

Explanation:
- `os.path.getsize()` returns the file size in bytes.
- If the size is zero, the file is empty.
- Otherwise, the program reads its content.

##Q23. Write a Python program that writes to a log file when an error occurs during file handling.

```python
import logging
logging.basicConfig(
    filename="file_errors.log",
    level=logging.ERROR,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
try:
    with open("missing.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    logging.exception("File handling error: File not found!")
    print("An error occurred. Check file_errors.log.")
except PermissionError:
    logging.exception("File handling error: Permission denied!")
    print("Permission denied. Check file_errors.log.")
except OSError:
    logging.exception("Unexpected file handling error!")
    print("File operation failed. Check file_errors.log.")
```

Explanation:
- The program attempts to open a file.
- If the file is missing or inaccessible, an exception occurs.
- `logging.exception()` records the error and traceback.
- The error details are saved in `file_errors.log`.
